```python
import pandas as pd
from pathlib import Path

data_path = Path(
    r"C:\Users\shubh\A Rain drop volume prediction project\Data\raindrop_volume_dataset.csv"
)

# Load the complete dataset
df = pd.read_csv(data_path)

print("Full dataset loaded!")
print("Rows and columns:", df.shape)
```

```python
print("Missing values:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nNumerical statistics:")
display(df.describe())
```

```python
# Count exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Total rows:", len(df))
print("Exact duplicate rows:", duplicate_count)
print("Duplicate percentage:", round(duplicate_count / len(df) * 100, 3), "%")

# Display some duplicate records
duplicates = df[df.duplicated(keep=False)]
display(duplicates.head(10))
```

```python
measurement_cols = [
    "DateTime",
    "Diameter_mm",
    "Velocity_m_s",
    "Volume_mm3",
    "Condition"
]

print(
    "Duplicates excluding SourceFile:",
    df.duplicated(subset=measurement_cols).sum()
)
```

```python
# Number of rows contributed by each source file
print("Rows per source file:")
display(df["SourceFile"].value_counts())

# Number of distinct source files
print("\nUnique source files:", df["SourceFile"].nunique())
```

```python
# Check how many rows remain after keeping one copy of each exact row
df_unique = df.drop_duplicates()

print("Original rows:", len(df))
print("Rows after removing exact duplicates:", len(df_unique))
print("Rows removed:", len(df) - len(df_unique))
print("Remaining duplicate rows:", df_unique.duplicated().sum())
```

```python
# Check duplicate rows separately for each source file

for source, group in df.groupby("SourceFile"):
    print("\nSource file:", source)
    print("Total rows:", len(group))
    print("Exact duplicates:", group.duplicated().sum())
    print(
        "Unique rows:",
        len(group.drop_duplicates())
    )
```

```python
# Check repeated measurement combinations
measurement_cols = [
    "DateTime",
    "Diameter_mm",
    "Velocity_m_s"
]

print("Repeated measurement combinations:")
display(
    df[df.duplicated(subset=measurement_cols, keep=False)]
    .sort_values(measurement_cols)
    .head(20)
)
```

```python
# Count how many distinct timestamps exist
print("Unique timestamps:", df["DateTime"].nunique())

# Count unique diameter and velocity combinations
print(
    "Unique diameter-velocity pairs:",
    df[["Diameter_mm", "Velocity_m_s"]].drop_duplicates().shape[0]
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["Diameter_mm"], bins=50)
plt.xlabel("Raindrop Diameter (mm)")
plt.ylabel("Frequency")
plt.title("Distribution of Raindrop Diameters")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Volume_mm3"], bins=50)
plt.xlabel("Raindrop Volume (mm³)")
plt.ylabel("Frequency")
plt.title("Distribution of Raindrop Volumes")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Diameter_mm"].sample(5000, random_state=42),
    df["Volume_mm3"].sample(5000, random_state=42),
    alpha=0.4
)
plt.xlabel("Diameter (mm)")
plt.ylabel("Volume (mm³)")
plt.title("Diameter vs Volume")
plt.tight_layout()
plt.show()
```

```python
sample_df = df.sample(5000, random_state=42)

plt.figure(figsize=(8, 5))
plt.scatter(
    sample_df["Diameter_mm"],
    sample_df["Volume_mm3"],
    alpha=0.4
)
plt.xlabel("Diameter (mm)")
plt.ylabel("Volume (mm³)")
plt.title("Diameter vs Volume")
plt.tight_layout()
plt.show()
```

1. **Diameter vs Volume:** The relationship between raindrop diameter and volume is nonlinear. Volume increases with the cube of diameter.
2. **Volume Distribution:** The distribution plots help understand the frequency and spread of raindrop volumes in the dataset.
3. **Data Quality:** Missing values and duplicate records were examined during data exploration.
